# XSimGCL: An Extremely Simple Graph Contrastive Learning for Recommendation

## Paper

**XSimGCL: An Extremely Simple Graph Contrastive Learning for Recommendation**

Published at **CIKM 2023**

## Objective

Implement XSimGCL from scratch using the MovieLens-100K dataset.

Unlike SimGCL — which injects noise at every propagation layer — XSimGCL removes perturbation from propagation entirely and instead perturbs only the final embedding. This drastically reduces computation while preserving (and often improving) representation quality.

## Pipeline

1. Load MovieLens Dataset
2. Build Bipartite Graph
3. Normalize Graph
4. Build LightGCN Encoder
5. Generate Perturbed Embeddings (once, after propagation)
6. Compute BPR Loss
7. Compute InfoNCE Loss
8. Joint Optimization
9. Evaluate Recommendation Performance

```python
NUM_USERS = data["user_id"].nunique()

NUM_ITEMS = data["movie_id"].nunique()

print(NUM_USERS)

print(NUM_ITEMS)
```

```python
interaction_matrix = sp.coo_matrix(

    (

        np.ones(

            len(train_df)

        ),

        (

            train_df["user_id"],

            train_df["movie_id"]

        )

    ),

    shape=(

        NUM_USERS,

        NUM_ITEMS

    )

)

interaction_matrix
```

```python
num_nodes = NUM_USERS + NUM_ITEMS

adjacency = sp.dok_matrix(

    (

        num_nodes,

        num_nodes

    ),

    dtype=np.float32

)

adjacency = adjacency.tolil()

adjacency[

    :NUM_USERS,

    NUM_USERS:

] = interaction_matrix

adjacency[

    NUM_USERS:,

    :NUM_USERS

] = interaction_matrix.T

adjacency = adjacency.tocsr()

adjacency
```

```python
degree = np.array(

    adjacency.sum(axis=1)

).flatten()

degree_inv = np.power(

    degree,

    -0.5

)

degree_inv[

    np.isinf(

        degree_inv

    )

] = 0

D = sp.diags(

    degree_inv

)

normalized_graph = (

    D

    @

    adjacency

    @

    D

)
```

```python
normalized_graph = normalized_graph.tocoo()

indices = torch.LongTensor(

    np.vstack(

        (

            normalized_graph.row,

            normalized_graph.col

        )

    )

)

values = torch.FloatTensor(

    normalized_graph.data

)

graph = torch.sparse_coo_tensor(

    indices,

    values,

    normalized_graph.shape

).coalesce()
```

```python
class BPRDataset(Dataset):

    def __init__(

        self,

        dataframe,

        num_items

    ):

        self.users = dataframe["user_id"].values

        self.pos_items = dataframe["movie_id"].values

        self.num_items = num_items

        self.user_history = (

            dataframe

            .groupby("user_id")["movie_id"]

            .apply(set)

            .to_dict()

        )

    def __len__(self):

        return len(self.users)

    def __getitem__(

        self,

        idx

    ):

        user = self.users[idx]

        positive = self.pos_items[idx]

        while True:

            negative = np.random.randint(

                self.num_items

            )

            if negative not in self.user_history[user]:

                break

        return (

            torch.LongTensor([user]).squeeze(),

            torch.LongTensor([positive]).squeeze(),

            torch.LongTensor([negative]).squeeze()

        )
```

```python
class XSimGCL(nn.Module):

    def __init__(

        self,

        num_users,

        num_items,

        embedding_dim=64,

        num_layers=3,

        eps=0.1

    ):

        super().__init__()

        self.num_users = num_users

        self.num_items = num_items

        self.num_layers = num_layers

        self.eps = eps

        self.user_embedding = nn.Embedding(

            num_users,

            embedding_dim

        )

        self.item_embedding = nn.Embedding(

            num_items,

            embedding_dim

        )

        nn.init.xavier_uniform_(

            self.user_embedding.weight

        )

        nn.init.xavier_uniform_(

            self.item_embedding.weight

        )

    def perturb(

        self,

        embedding

    ):

        noise = torch.rand_like(

            embedding

        )

        noise = F.normalize(

            noise,

            dim=1

        )

        perturbation = (

            torch.sign(

                embedding

            )

            *

            noise

            *

            self.eps

        )

        return embedding + perturbation
    
    
    def propagate(self,graph):

        embeddings = torch.cat(

            [self.user_embedding.weight,self.item_embedding.weight],dim=0
)

        all_embeddings = [embeddings]

        for _ in range(

            self.num_layers

        ):

            embeddings = torch.sparse.mm(

                graph,

                embeddings

            )


            all_embeddings.append(

                embeddings

            )

        final_embedding = torch.stack(

            all_embeddings,

            dim=1

        ).mean(dim=1)

        return final_embedding
    
    def forward(self, graph):

        clean_embedding = self.propagate(graph)

        view1 = self.perturb(clean_embedding)

        view2 = self.perturb(clean_embedding)

        return clean_embedding, view1, view2    
```

```python
model = XSimGCL(

    NUM_USERS,

    NUM_ITEMS,

    embedding_dim=64,

    num_layers=3,

    eps=0.1

)
model
```

```python
def normalize(

    embedding

):

    return F.normalize(

        embedding,

        dim=1

    )
```

```python
def similarity(

    x,

    y

):

    return torch.matmul(

        normalize(x),

        normalize(y).T

    )
```

```python
def info_nce_loss(

    view1,

    view2,

    temperature=0.2

):

    view1 = normalize(

        view1

    )

    view2 = normalize(

        view2

    )

    similarity_matrix = torch.matmul(

        view1,

        view2.T

    )

    similarity_matrix /= temperature

    labels = torch.arange(

        similarity_matrix.size(0),

        device=view1.device

    )

    loss = F.cross_entropy(

        similarity_matrix,

        labels

    )

    return loss
```

```python
loss = info_nce_loss(

    view1,

    view2

)

print(loss)
```

```python
def bpr_loss(

    positive,

    negative

):

    return -torch.mean(

        F.logsigmoid(

            positive

            -

            negative

        )

    )
```

```python
def compute_scores(

    embeddings,

    users,

    positive_items,

    negative_items

):

    user_embedding = embeddings[

        users

    ]

    positive_embedding = embeddings[

        NUM_USERS +

        positive_items

    ]

    negative_embedding = embeddings[

        NUM_USERS +

        negative_items

    ]

    positive_scores = (

        user_embedding *

        positive_embedding

    ).sum(dim=1)

    negative_scores = (

        user_embedding *

        negative_embedding

    ).sum(dim=1)

    return (

        positive_scores,

        negative_scores

    )
```

```python
def simgcl_loss(

    positive_scores,

    negative_scores,

    view1,

    view2,

    lambda_ssl=0.1

):

    ranking_loss = bpr_loss(

        positive_scores,

        negative_scores

    )

    contrastive_loss = info_nce_loss(

        view1,

        view2

    )

    return (

        ranking_loss

        +

        lambda_ssl *

        contrastive_loss

    )
```

```python
optimizer = torch.optim.Adam(

    model.parameters(),

    lr=0.001,

    weight_decay=1e-4

)
```

```python
def train_model(

    model,

    train_loader,

    graph,

    optimizer,

    epochs=30

):

    history = []

    print(

        "Training XSimGCL..."

    )

    for epoch in range(

        epochs

    ):

        model.train()

        total_loss = 0

        for (

            users,

            positive_items,

            negative_items

        ) in train_loader:

            optimizer.zero_grad()

            (

                clean,

                view1,

                view2

            ) = model(

                graph

            )

            (

                positive_scores,

                negative_scores

            ) = compute_scores(

                clean,

                users,

                positive_items,

                negative_items

            )

            loss = simgcl_loss(

                positive_scores,

                negative_scores,

                view1,

                view2

            )

            loss.backward()

            optimizer.step()

            total_loss += loss.item()

        epoch_loss = total_loss / len(

            train_loader

        )

        history.append(

            epoch_loss

        )

        print(

            f"Epoch {epoch+1:2d}/{epochs} | Loss = {epoch_loss:.4f}"

        )

    return history
```

```python
history = train_model(

    model,

    train_loader,

    graph,

    optimizer,

    epochs=10

)
```

```python
def recommend(

    user_id,

    top_k=10

):

    user_embedding = clean_embedding[

        user_id

    ]

    item_embeddings = clean_embedding[

        NUM_USERS:

    ]

    scores = torch.matmul(

        item_embeddings,

        user_embedding

    )

    _, indices = torch.topk(

        scores,

        top_k

    )

    return indices.cpu().numpy()
```

```python
def precision_at_k(

    prediction_df,

    k=10

):

    precisions = []

    grouped = prediction_df.groupby(

        "user_id"

    )

    for (

        _,

        group

    ) in grouped:

        top_k = group.sort_values(

            "probability",

            ascending=False

        ).head(k)

        precisions.append(

            top_k["label"].mean()

        )

    return np.mean(

        precisions

    )
```

```python
def recall_at_k(

    prediction_df,

    k=10

):

    recalls = []

    grouped = prediction_df.groupby(

        "user_id"

    )

    for (

        _,

        group

    ) in grouped:

        relevant = group["label"].sum()

        if relevant == 0:

            continue

        top_k = group.sort_values(

            "probability",

            ascending=False

        ).head(k)

        hits = top_k["label"].sum()

        recalls.append(

            hits / relevant

        )

    return np.mean(

        recalls

    )
```

```python
def hit_rate_at_k(

    prediction_df,

    k=10

):

    hits = []

    grouped = prediction_df.groupby(

        "user_id"

    )

    for (

        _,

        group

    ) in grouped:

        top_k = group.sort_values(

            "probability",

            ascending=False

        ).head(k)

        hits.append(

            top_k["label"].sum() > 0

        )

    return np.mean(

        hits

    )
```

```python
precision = precision_at_k(

    prediction_df,

    k=10

)

recall = recall_at_k(

    prediction_df,

    k=10

)

hitrate = hit_rate_at_k(

    prediction_df,

    k=10

)
```

# Conclusion

In this notebook, we implemented **XSimGCL**, an extremely simple graph contrastive learning framework for recommendation.

Key observations:

- XSimGCL keeps the same LightGCN encoder as SimGCL, with no architectural change to the graph propagation.
- The key difference from SimGCL is that perturbation is removed from inside the propagation loop. Instead, XSimGCL computes the clean final embedding once, then applies the noise function twice to produce two contrastive views.
- This means the graph is propagated once per batch instead of three times, which roughly halves the wall-clock training cost of SimGCL.
- The InfoNCE objective, BPR ranking objective, and evaluation pipeline are otherwise unchanged.
- The "X" stands for *eXtremely simple*: XSimGCL demonstrates that the contrastive benefit of SimGCL comes primarily from the final-embedding perturbation, not from per-layer noise.

Compared with SimGCL, XSimGCL achieves comparable — and often slightly better — representation quality while significantly reducing computation by removing unnecessary perturbation inside propagation.
